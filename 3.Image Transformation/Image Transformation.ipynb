**Image Translation**

```python
import numpy as np
import matplotlib.pyplot as plt
import cv2
import math

img = cv2.cvtColor(cv2.imread("birb.jpeg"), cv2.COLOR_BGR2RGB)

def translate_image(img, bx, by):
    height, width = img.shape[:2]
    translated_img = np.zeros_like(img)

    for y in range(height):
        for x in range(width):
            new_x = x + bx
            new_y = y + by

            if 0 <= new_x < width and 0 <= new_y < height:
                translated_img[new_y, new_x] = img[y, x]

    return translated_img

translated_img = translate_image(img, 200, 0)

def rotate_image(img, angle):
    height, width = img.shape[:2]
    angle_rad = math.radians(angle)
    center_x, center_y = width // 2, height // 2
    rotated_img = np.zeros_like(img)

    for y in range(height):
        for x in range(width):
            tx = x - center_x
            ty = y - center_y
            new_x = int(tx * math.cos(angle_rad) - ty * math.sin(angle_rad) + center_x)
            new_y = int(tx * math.sin(angle_rad) + ty * math.cos(angle_rad) + center_y)

            if 0 <= new_x < width and 0 <= new_y < height:
                rotated_img[y, x] = img[new_y, new_x]

    return rotated_img

rotated_img = rotate_image(img, 90)

plt.subplot(1, 2, 1)
plt.imshow(img)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(translated_img)
plt.title("Translated Image")
plt.axis("off")

plt.show()
```

**Image Rotation**

```python
def rotate_image(img, angle):
    height, width = img.shape[:2]
    angle_rad = math.radians(angle)
    center_x, center_y = width // 2, height // 2
    rotated_img = np.zeros_like(img)

    for y in range(height):
        for x in range(width):
            tx = x - center_x
            ty = y - center_y
            new_x = int(tx * math.cos(angle_rad) - ty * math.sin(angle_rad) + center_x)
            new_y = int(tx * math.sin(angle_rad) + ty * math.cos(angle_rad) + center_y)

            if 0 <= new_x < width and 0 <= new_y < height:
                rotated_img[y, x] = img[new_y, new_x]

    return rotated_img

rotated_img = rotate_image(img, 90)

plt.subplot(1, 2, 1)
plt.imshow(img)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(rotated_img)
plt.title("Rotated Image")
plt.axis("off")

plt.show()
```

**Image Scaling**

```python
def scale_image(img, scale_x, scale_y):
    height, width = img.shape[:2]
    new_width = int(width * scale_x)
    new_height = int(height * scale_y)
    scaled_img = np.zeros((new_height, new_width, 3), dtype=np.uint8)

    for y_new in range(new_height):
        for x_new in range(new_width):
            x_old = int(x_new / scale_x)
            y_old = int(y_new / scale_y)
            if x_old < width and y_old < height:
                scaled_img[y_new, x_new] = img[y_old, x_old]

    return scaled_img

scaled_img = scale_image(img, 2, 2)

plt.subplot(1, 2, 1)
plt.imshow(img)
plt.title("Original Image")
# plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(scaled_img)
plt.title("Scaled Image")
# plt.axis("off")

plt.show()
```

**Image Shearing**

```python
def shear_image(img, shear_factor_x=0, shear_factor_y=0):
    height, width = img.shape[:2]
    new_width = width + int(abs(shear_factor_x) * height)
    new_height = height + int(abs(shear_factor_y) * width)
    sheared_img = np.zeros((new_height, new_width, 3), dtype=np.uint8)

    for y in range(height):
        for x in range(width):
            new_x = int(x + shear_factor_x * y)
            new_y = int(y + shear_factor_y * x)
            if 0 <= new_x < new_width and 0 <= new_y < new_height:
                sheared_img[new_y, new_x] = img[y, x]

    return sheared_img

sheared_img_x = shear_image(img, shear_factor_x=0.5)
sheared_img_y = shear_image(img, shear_factor_y=0.5)

plt.figure(figsize=(10, 8))

plt.subplot(2, 2, 1)
plt.imshow(img)
plt.title("Original Image")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(sheared_img_x)
plt.title("Sheared Image X")
plt.axis("off")

plt.subplot(2, 2, 3)
plt.imshow(sheared_img_y)
plt.title("Sheared Image Y")
plt.axis("off")

plt.show()
```

**Image Reflection**

```python
def reflect_image(img, axis='horizontal'):
    height, width = img.shape[:2]
    reflected_img = np.zeros_like(img)

    if axis == 'horizontal':
        for y in range(height):
            for x in range(width):
                new_x = width - x - 1
                reflected_img[y, new_x] = img[y, x]
    elif axis == 'vertical':
        for y in range(height):
            for x in range(width):
                new_y = height - y - 1
                reflected_img[new_y, x] = img[y, x]

    return reflected_img

reflected_img_horizontal = reflect_image(img, axis='horizontal')
reflected_img_vertical = reflect_image(img, axis='vertical')

plt.figure(figsize=(10, 8))

plt.subplot(2, 2, 1)
plt.imshow(img)
plt.title("Original Image")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(reflected_img_horizontal)
plt.title("Horizontal Reflected Image")
plt.axis("off")

plt.subplot(2, 2, 3)
plt.imshow(reflected_img_vertical)
plt.title("Vertical Reflected Image")
plt.axis("off")

plt.show()
```
