**Image in HSV**

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
image = cv2.imread('th_old.png')
image_hsv = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
image_hsv = cv2.cvtColor(image_hsv, cv2.COLOR_RGB2HSV)
h_array=image_hsv
h_array = (h_array * 255).astype(np.uint8)
h_array[:,:,0]=22*h_array[:,:,0]
image_h = cv2.cvtColor(h_array, cv2.COLOR_HSV2RGB)
s_array=image_hsv
s_array = (s_array * 255).astype(np.uint8)
s_array[:,:,1]=250*s_array[:,:,1]
image_s = cv2.cvtColor(s_array, cv2.COLOR_BGR2RGB)
v_array=image_hsv
v_array = (v_array * 25).astype(np.uint8)
v_array[:,:,2]=215*v_array[:,:,2]
image_v = cv2.cvtColor(v_array, cv2.COLOR_BGR2RGB)
plt.figure(figsize=(50,50))
a = np.hstack((image_hsv,image_h,image_s,image_v))
plt.imshow(a)
plt.axis('off')
plt.show()
```

**Image in CMY**

```python
from PIL import Image
import numpy as np
import matplotlib.pyplot as plt
image = Image.open('th_old.png')
image = image.convert('RGB')
rgb_array = np.array(image) / 255.0
cmy_array = 1 - rgb_array
cmy_array = (cmy_array * 255).astype(np.uint8)
c_array = cmy_array
c_array = (c_array * 255).astype(np.uint8)
c_array[:,:,0] = 1
m_array = cmy_array
m_array = (m_array * 255).astype(np.uint8)
m_array[:,:,1] = 1
y_array = cmy_array
y_array = (y_array * 255).astype(np.uint8)
y_array[:,:,2] = 1
cm_array = cmy_array
cm_array = (cm_array * 255).astype(np.uint8)
cm_array[:,:,0:2] = 1
my_array = cmy_array
my_array = (my_array * 255).astype(np.uint8)
my_array[:,:,1:3] = 1
cy_array = cmy_array
cy_array = (cy_array * 255).astype(np.uint8)
cy_array[:,:,0:3:2] = 1
plt.figure(figsize=(50,50))
a = np.hstack((cmy_array, c_array, m_array, y_array))
plt.imshow(a)
plt.axis('off')
plt.show()
plt.figure(figsize=(50,50))
b = np.hstack((image,my_array, cy_array, cm_array))
plt.imshow(b)
plt.axis('off')
plt.show()
```
